```python
import pandas as pd
import glob

fichier_data = glob.glob("../data/raw/Turbine_Data_Kelmarsh_1_*.csv")[0]
fichier_status = glob.glob("../data/raw/Status_Kelmarsh_1_*.csv")[0]

print(fichier_data)
print(fichier_status)
```

```python
with open(fichier_data, encoding="utf-8", errors="replace") as f:
    for i in range(12):
        print(f.readline()[:300])
```

```python
colonnes = pd.read_csv(fichier_data, skiprows=9, nrows=0).columns
colonnes = [c.replace("# ", "", 1) if c.startswith("# ") else c for c in colonnes]
print(len(colonnes))
for i, c in enumerate(colonnes):
    print(i, c)
```

```python
cols_utiles = [
    "Date and time",
    "Wind speed (m/s)", "Wind speed Sensor 1 (m/s)", "Wind speed Sensor 2 (m/s)",
    "Power (kW)", "Potential power default PC (kW)", "Power, Maximum (kW)",
    "Rotor speed (RPM)", "Generator RPM (RPM)", "Blade angle (pitch position) A (°)",
    "Gear oil temperature (°C)", "Front bearing temperature (°C)", "Stator temperature 1 (°C)",
    "Data Availability", "Time-based System Avail.", "Lost Production Total (kWh)",
]

df = pd.read_csv(
    fichier_data,
    skiprows=9,
    usecols=lambda c: c.replace("# ", "", 1) in cols_utiles,
)
df.columns = [c.replace("# ", "", 1) for c in df.columns]
df["Date and time"] = pd.to_datetime(df["Date and time"])
df = df.set_index("Date and time")

print(df.shape)
df.head()
```

```python
print(df.index.min(), df.index.max())
print("Lignes :", len(df))
print("Timestamps uniques :", df.index.nunique())
print("Lignes entièrement dupliquées :", df.duplicated().sum())
print()
print(df.index.to_series().diff().value_counts().head(8))
print()
print(df.groupby(df.index.year).size())
```

```python
n_distincts = df.reset_index().drop_duplicates().shape[0]
print("Lignes distinctes (date + valeurs) :", n_distincts)
print("Timestamps uniques :", df.index.nunique())
```

```python
# Une date au hasard avec plusieurs versions distinctes
d = df.reset_index().drop_duplicates()
comptes = d.groupby("Date and time").size()
print(comptes.value_counts())   # combien de dates ont 1, 2, 3... versions

# Prendre la 1re date qui a 2 versions et voir quelles colonnes diffèrent
date_test = comptes[comptes == 2].index[0]
v = d[d["Date and time"] == date_test].set_index("Date and time").T
v.columns = ["version A", "version B"]
v["identique"] = (v["version A"] == v["version B"]) | (v["version A"].isna() & v["version B"].isna())
print(date_test)
print(v[~v["identique"]])
```

```python
df_propre = df.dropna(how="all")

print("Lignes après suppression des lignes vides :", len(df_propre))
print("Timestamps uniques :", df_propre.index.nunique())
print("Lignes distinctes (date + valeurs) :", df_propre.reset_index().drop_duplicates().shape[0])
```

```python
df = df_propre[~df_propre.index.duplicated(keep="first")].sort_index()

print(df.shape)
print(df.index.is_monotonic_increasing)
print(df.index.to_series().diff().value_counts())
```

```python
print(df.isna().sum())
print()
print((df.isna().mean() * 100).round(2).astype(str) + " %")
df.describe().T[["min", "mean", "max"]]
```

```python
# Capteur 1 : combien de valeurs exactement à 0 ?
print("Vent capteur 1 == 0 :", (df["Wind speed Sensor 1 (m/s)"] == 0).sum())

# Écart entre les deux capteurs (grand écart = un capteur suspect)
ecart = (df["Wind speed Sensor 1 (m/s)"] - df["Wind speed Sensor 2 (m/s)"]).abs()
print(ecart.describe())
```

```python
manquants = pd.DataFrame({
    "nombre": df.isna().sum(),
    "pourcentage": (df.isna().mean() * 100).round(2),
})
print(manquants)
```

```python
trous = df["Power (kW)"].isna()
print(df.loc[trous, "Data Availability"].value_counts())

# Regroupement en blocs consécutifs
bloc = (trous != trous.shift()).cumsum()
blocs = trous.groupby(bloc).agg(["first", "size"])
blocs = blocs[blocs["first"]]
print("Nombre de blocs de trous :", len(blocs))
print(blocs["size"].describe())
print(trous[trous].index[:10])
```

```python
ecart = (df["Wind speed (m/s)"] - df["Wind speed Sensor 2 (m/s)"]).abs()
print(ecart.describe())
print("Corrélation :", df["Wind speed (m/s)"].corr(df["Wind speed Sensor 2 (m/s)"]))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("kelmarsh1_2023_propre.csv", index_col=0, parse_dates=True)
print(df.shape)
```

```python
d = df[["Wind speed (m/s)", "Power (kW)"]].dropna()

plt.figure(figsize=(9, 5))
plt.scatter(d["Wind speed (m/s)"], d["Power (kW)"], s=2, alpha=0.2)
plt.xlabel("Vitesse du vent (m/s)")
plt.ylabel("Puissance (kW)")
plt.title("Kelmarsh 1 - 2023 : puissance en fonction du vent")
plt.grid(True)
plt.show()
```

```python
th = df[["Wind speed (m/s)", "Potential power default PC (kW)"]].dropna().copy()
th["classe"] = (th["Wind speed (m/s)"] // 0.5) * 0.5 + 0.25
courbe_th = th.groupby("classe")["Potential power default PC (kW)"].mean()

plt.figure(figsize=(9, 5))
plt.plot(courbe.index, courbe["mean"], label="Mesurée (moyenne par classe)")
plt.plot(courbe_th.index, courbe_th, "--", label="Théorique (default PC)")
plt.xlabel("Vitesse du vent (m/s)")
plt.ylabel("Puissance (kW)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

p = df[["Wind speed (m/s)", "Blade angle (pitch position) A (°)"]].dropna().copy()
p.columns = ["vent", "pitch"]

# Moyenne par classe de vent de 0,5 m/s
p["classe"] = (p["vent"] // 0.5) * 0.5 + 0.25
moy = p.groupby("classe")["pitch"].mean()

plt.figure(figsize=(9, 5))
plt.scatter(p["vent"], p["pitch"], s=2, alpha=0.15, label="Mesures 10 min")
plt.plot(moy.index, moy.values, color="red", linewidth=2, label="Moyenne par classe")
plt.xlabel("Vitesse du vent (m/s)")
plt.ylabel("Angle de pitch, pale A (°)")
plt.title("Kelmarsh 1 - 2023 : pitch en fonction du vent")
plt.ylim(-5, 40)   # coupe les points à ~90° (turbine en drapeau) pour voir la zone utile
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

q = df[["Power (kW)", "Blade angle (pitch position) A (°)"]].dropna()
q = q[q["Power (kW)"] > 0]   # on garde seulement la turbine en production

plt.figure(figsize=(9, 5))
plt.scatter(q["Power (kW)"], q["Blade angle (pitch position) A (°)"], s=2, alpha=0.15)
plt.xlabel("Puissance récupérée (kW)")
plt.ylabel("Angle de pitch, pale A (°)")
plt.title("Kelmarsh 1 - 2023 : pitch en fonction de la puissance")
plt.ylim(-2, 30)
plt.grid(True)
plt.show()
```

```python
import numpy as np

c = df[["Wind speed (m/s)", "Power (kW)"]].dropna().copy()
c = c[(c["Wind speed (m/s)"] > 3) & (c["Power (kW)"] > 0)]
rho, A = 1.225, np.pi * 46**2
c["Cp"] = c["Power (kW)"] * 1000 / (0.5 * rho * A * c["Wind speed (m/s)"]**3)

print(c["Cp"].describe())
print("Quantiles 1 % et 99 % :", c["Cp"].quantile([0.01, 0.99]).round(3).tolist())
print("Lignes avec Cp > 0,593 (impossible) :", (c["Cp"] > 0.593).sum())
```

```python
c2 = c.join(df["Blade angle (pitch position) A (°)"], how="left")
suspect = c2[(c2["Wind speed (m/s)"].between(4, 10)) &
             (c2["Blade angle (pitch position) A (°)"] < 2) &
             (c2["Cp"] < 0.3)]
print("Lignes suspectes (Cp < 0,3 en zone optimale) :", len(suspect))
print(suspect.index.to_series().dt.date.value_counts().head(10))
```
